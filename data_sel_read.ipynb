### Selecting rows and columns 

loc selects using row/column labels. iloc selects using integer positions.


```python

df.loc[0, 'restaurant name']
```

```python

df.loc[0:4, ['restaurant name', 'area', 'rate (out of 5)']]
```

```python

df.iloc[0:5, 0:4]
```
