DATA TRANSFORMATION


```python
import pandas as pd
import numpy as np

# Read the original dataset
df = pd.read_csv("zomato.csv")

# Display first 5 rows
df.head()
```

```python
# Work on a copy of the original data
zomato = df.copy()

# Check whether the two unnamed columns contain identical values
print("Are the two index columns identical?",
      zomato['Unnamed: 0.1'].equals(zomato['Unnamed: 0']))
```

```python
# Display remaining columns
zomato.head()
```

```python
# Drop unnecessary columns
zomato = zomato.drop(
    ['Unnamed: 0.1', 'Unnamed: 0'],
    axis=1,
    errors="ignore"
)

zomato.head()
```

```python
print(zomato.columns)
```

RENAME THE COLUMNS

```python
zomato = zomato.rename(columns={
    'restaurant name': 'restaurant_name',
    'restaurant type': 'restaurant_type',
    'rate (out of 5)': 'rating',
    'num of ratings': 'num_ratings',
    'avg cost (two people)': 'average_cost_for_two',
    'online_order': 'online_order',
    'table booking': 'table_booking',
    'cuisines type': 'cuisines',
    'area': 'area',
    'local address': 'local_address'
})

zomato.head()
```

ADDING NEW COLUMNS

```python
# Count the number of cuisines listed for each restaurant
zomato['cuisine_count'] = zomato['cuisines'].apply(lambda x: len(x.split(',')))

# Display selected columns
zomato[['restaurant_name', 'cuisines', 'cuisine_count']].head(10)
```

Handling Null Values

```python
print(zomato.isnull().sum())
```

```python
rating_mean = zomato['rating'].mean()
cost_mean = zomato['average_cost_for_two'].mean()

print("Mean restaurant rating:", rating_mean)
print("Mean average cost for two:", cost_mean)
```

```python
# Filling , missing rating values using the mean rating
zomato['rating'] = zomato['rating'].fillna(rating_mean)
```

```python
# Filling , missing average-cost values using the mean average cost
zomato['average_cost_for_two'] = zomato['average_cost_for_two'].fillna(cost_mean)
```

```python
print(zomato.isnull().sum())
```

```python
print("Rows after null handling:", zomato.shape[0])
```
