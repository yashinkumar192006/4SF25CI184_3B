```python
import pandas as pd


zomato = pd.read_csv('zomato.csv')
```

```python
restaurants_by_area = zomato['area'].value_counts()

print("Top 10 areas by number of restaurants:")
print(restaurants_by_area.head(10))
```

```python
print(zomato.columns.tolist())
```

```python
zomato.groupby("area")["rate (out of 5)"].mean()
```

```python
restaurant_type_insights = zomato.groupby('restaurant type').agg(
    number_of_restaurants=('restaurant name', 'count'),
    average_rating=('rate (out of 5)', 'mean'),
    average_cost=('avg cost (two people)', 'mean')
)


restaurant_type_insights_30 = restaurant_type_insights[
    restaurant_type_insights['number_of_restaurants'] >= 30
]

restaurant_type_insights_30.sort_values(
    'average_rating', ascending=False
).round(2)


restaurant_type_insights_30 = restaurant_type_insights[
    restaurant_type_insights['number_of_restaurants'] >= 30
]

restaurant_type_insights_30.sort_values(
    'average_rating', ascending=False
).round(2)
```

```python

online_order_insights = zomato.groupby('online_order').agg(
    number_of_restaurants=('restaurant name', 'count'),
    average_rating=('rate (out of 5)', 'mean'),
    average_cost=('avg cost (two people)', 'mean')
)

online_order_insights.round(2)
```
